# **Course Enrollment**

1. Create the Student DataBase

```r
students <- data.frame(
  Student_ID = c("S001", "S002", "S003", "S004", "S005", "S006", "S007", "S008"),
  Programming_Grade = c(88, 72, 91, 78, 65, 85, 74, 82),
  Math_Grade = c(85, 70, 88, 80, 68, 90, 73, 79),
  Attendance = c(92, 80, 95, 88, 70, 91, 76, 85),
  Prerequisite = c("Yes", "No", "Yes", "Yes", "No", "Yes", "No", "Yes")
)

students
```

2. Explore the Data

```r
summary(students)
nrow(students)
ncol(students)


max(students$Programming_Grade)
max(students$Math_Grade)
max(students$Attendance)
max(students$Prerequisite)

cat("\n")

max(students$Math_Grade)
min(students$Math_Grade)
ave(students$Math_Grade)
mean(students$Math_Grade)
sum(students$Math_Grade)

cat("\n")

max(students$Attendance)
min(students$Attendance)
ave(students$Attendance)
mean(students$Attendance)
sum(students$Attendance)

```

6. Extract Student Information

```r
programming = student$Programming_Grade
math = student$Math_Grade
attendance = student$Attendance
prerequisite = student$Prerequisite

cat("Programming Grade: ", programming)
cat("\nMath Grade: ", math)
cat("\nAttendace: ", attendance)
cat("\nPrerequisite: ", prerequisite)

```

7. Implement the Rules

Rule1 - Eligible: Programming Grade >= 80 AND Math Grade >= 75 AND Attendance >= 80% AND Prerequisite = Yes.

Rule 2 - Conditionally Eligible: Programming Grade >= 75 AND Math Grade >= 70 AND Attendance >= 75%, but Rule 1 is not satisfied.

Rule 3 - Not Eligible: If neither Rule 1 nor Rule 2 is satisfied, classify the student as Not Eligible.


```r
if (programming >= 80 && math >= 75 && attendance >= 80 && prerequisite == "Yes") {
  classification <- "Eligible"
} else if (programming >= 75 && math >= 70 && attendance >= 75) {
  classification <- "Conditionally Eligible"
} else {
  classification <- "Not Eligible"
}

cat("Student Classification: ", classification)
```

8. Apply the Functions to All students

```r
classify_student <- function(student_row) {
  programming <- student_row['Programming_Grade']
  math <- student_row['Math_Grade']
  attendance <- student_row['Attendance']
  prerequisite <- student_row['Prerequisite']

  if (programming >= 80 && math >= 75 && attendance >= 80 && prereq == "Yes") {
    return("Eligible")
  } else if (programming >= 75 && math >= 70 && attendance >= 75) {
    return("Conditionally Eligible")
  } else {
    return("Not Eligible")
  }
}

students$Classification <- apply(students, 1, classify_student)
students
```

9. Create Recommend Rules

```r
recommend_student <- function(classification_status) {
  if (classification_status == "Eligible") {
    return("Student may enroll")
  } else if (classification_status == "Conditionally Eligible") {
    return("Consult Academic adviser")
  } else {
    return("Complete requirements before enrollment")
  }
}

students$Recommendation <- sapply(students$Classification, recommend_student)
students
```
