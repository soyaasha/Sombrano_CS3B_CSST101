5. Connect Arguments and Evidences



```r
knowledge_base <- merge(
  arguments,
  evidence,
  by = "Argument_ID"
)
knowledge_base
```

6. Convert Evidences Strength into a Score



```r
get_score <- function(strength){
  if (strength == "Strong"){
    return(3)
  } else if (strength == "Moderate"){
    return(2)
  } else if (strength == "Weak"){
    return(1)
  } else{
    return(0)
  }
}

get_score("Strong")
```

7. Add Evidence Score

```r
knowledge_base$Evidence_Score <- sapply(
  knowledge_base$Strength,
  get_score
)
knowledge_base
```

8. Classify the Evidence Sdupport

```r
classify_support <- function(score){

  if (score == 3){
    return("Strongly Supported")
  } else if (score == 2){
    return("Moderately Supported")
  } else if (score == 1){
    return("Weakly Supported")
  } else{
    return("No Support")
  }
}

classify_support(2)
```

9. Search by Topic

```r
search_topic<- function(topic){
  result <- subset(
  knowledge_base,
  Topic == topic
  )

  if(nrow(result) == 0)(
    cat("Topic not found.\n")
  )else(
    print(result)
  )
}
```
