3. Numerical operators

```r
x <- 5
y <- 3

add <- x + y
sub <- x - y
mul <- x * y
div <- x / y

print(add)
print(sub)
print(mul)
print(div)


cat("+", add)
cat("-", sub)
cat("*", mul)
cat("/", div)
```

8. Vectors



```r
grades <- c(90, 94, 78, 83, 92)

grades

cat("\nMean: ",mean(grades))
cat("\nMax: ",max(grades))
cat("\nMin: ",min(grades))
cat("\nSum: ",sum(grades))
cat("\nLength: ",length(grades))
```
